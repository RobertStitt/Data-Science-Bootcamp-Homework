### Exercise 1: Sigmoid Function Implementation

Implement the sigmoid function from scratch and visualize it.

```python
import numpy as np
import matplotlib.pyplot as plt

# Implement the sigmoid function
def sigmoid(x):
    """
    Compute the sigmoid function for input x

    Formula: σ(x) = 1 / (1 + e^(-x))

    Parameters:
    x: input value or array

    Returns:
    sigmoid of x
    """
    return 1 / (1 + np.exp(-x))

# Generate x values from -10 to 10
x_values = np.linspace(-10, 10, 100)

# Compute sigmoid values for x_values
y_values = sigmoid(x_values)

# Create a plot of the sigmoid function
plt.figure(figsize=(12, 7))
plt.plot(x_values, y_values, 'b-', linewidth=3, label='σ(x) = 1/(1+e^(-x))')

# Add horizontal lines at y=0, 0.5, 1
plt.axhline(y=0, color='gray', linestyle='--', alpha=0.5, linewidth=1)
plt.axhline(y=0.5, color='red', linestyle='--', alpha=0.7, linewidth=1.5, label='y = 0.5 (decision boundary)')
plt.axhline(y=1, color='gray', linestyle='--', alpha=0.5, linewidth=1)

# Add vertical line at x=0
plt.axvline(x=0, color='red', linestyle='--', alpha=0.7, linewidth=1.5, label='x = 0')

# Mark the point (0, 0.5)
plt.plot(0, 0.5, 'ro', markersize=12, label='(0, 0.5)', zorder=5)

# Annotations for key properties
plt.annotate('Approaches 1\nas x → ∞', xy=(7, 0.95), fontsize=11,
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
plt.annotate('Approaches 0\nas x → -∞', xy=(-7, 0.05), fontsize=11,
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
plt.annotate('Inflection point\nat (0, 0.5)', xy=(0, 0.5), xytext=(2, 0.3),
            arrowprops=dict(arrowstyle='->', color='black', lw=1.5),
            fontsize=11, bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.8))

plt.title('Sigmoid Function σ(x) = 1/(1+e^(-x))', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('x', fontsize=13, fontweight='bold')
plt.ylabel('σ(x)', fontsize=13, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend(fontsize=11, loc='upper left')
plt.ylim(-0.1, 1.1)
plt.tight_layout()
plt.savefig('sigmoid_function.png', dpi=150, bbox_inches='tight')
plt.show()

# Test your implementation with specific values
test_values = [-5, -2, 0, 2, 5]
print("="*60)
print("SIGMOID FUNCTION TEST")
print("="*60)
print(f"{'x':<10} {'σ(x)':<15} {'Interpretation'}")
print("-"*60)
for val in test_values:
    sig_val = sigmoid(val)

    # Interpretation
    if sig_val < 0.3:
        interpretation = "Strong negative class"
    elif sig_val < 0.5:
        interpretation = "Weak negative class"
    elif sig_val == 0.5:
        interpretation = "Decision boundary"
    elif sig_val < 0.7:
        interpretation = "Weak positive class"
    else:
        interpretation = "Strong positive class"

    print(f"{val:<10} {sig_val:<15.8f} {interpretation}")

print("\n" + "="*60)
print("KEY PROPERTIES OF SIGMOID FUNCTION")
print("="*60)
print("1. Range: (0, 1) - output is always between 0 and 1")
print("2. Domain: (-∞, +∞) - accepts any real number")
print("3. σ(0) = 0.5 - midpoint/decision boundary")
print("4. Symmetric: σ(-x) = 1 - σ(x)")
print("5. Monotonically increasing - always goes up")
print("6. S-shaped curve - smooth transition")

# Additional visualizations
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot 1: Sigmoid with different perspectives
ax = axes[0, 0]
x_range = np.linspace(-10, 10, 1000)
y_range = sigmoid(x_range)
ax.plot(x_range, y_range, 'b-', linewidth=2.5)
ax.fill_between(x_range, 0, y_range, alpha=0.3, color='blue')
ax.axhline(y=0.5, color='red', linestyle='--', alpha=0.7, linewidth=1.5)
ax.axvline(x=0, color='red', linestyle='--', alpha=0.7, linewidth=1.5)
ax.set_title('Sigmoid Function with Shaded Area', fontsize=12, fontweight='bold')
ax.set_xlabel('x', fontsize=11)
ax.set_ylabel('σ(x)', fontsize=11)
ax.grid(True, alpha=0.3)
ax.set_ylim(-0.05, 1.05)

# Plot 2: Derivative of sigmoid
ax = axes[0, 1]
derivative = y_range * (1 - y_range)  # σ'(x) = σ(x)(1 - σ(x))
ax.plot(x_range, derivative, 'g-', linewidth=2.5, label="σ'(x) = σ(x)(1-σ(x))")
ax.axvline(x=0, color='red', linestyle='--', alpha=0.7, linewidth=1.5)
ax.plot(0, 0.25, 'ro', markersize=10, label='Maximum at (0, 0.25)', zorder=5)
ax.set_title("Derivative of Sigmoid σ'(x)", fontsize=12, fontweight='bold')
ax.set_xlabel('x', fontsize=11)
ax.set_ylabel("σ'(x)", fontsize=11)
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3)

# Plot 3: Sigmoid for classification
ax = axes[1, 0]
ax.plot(x_range, y_range, 'b-', linewidth=2.5)
ax.fill_between(x_range, 0, 0.5, where=(y_range < 0.5), alpha=0.3,
                color='red', label='Class 0 (Negative)')
ax.fill_between(x_range, 0.5, 1, where=(y_range >= 0.5), alpha=0.3,
                color='green', label='Class 1 (Positive)')
ax.axhline(y=0.5, color='black', linestyle='-', linewidth=2, label='Decision Boundary')
ax.set_title('Sigmoid in Binary Classification', fontsize=12, fontweight='bold')
ax.set_xlabel('x (logit/score)', fontsize=11)
ax.set_ylabel('P(y=1|x) - Probability', fontsize=11)
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3)
ax.set_ylim(-0.05, 1.05)

# Plot 4: Comparison with step function
ax = axes[1, 1]
ax.plot(x_range, y_range, 'b-', linewidth=2.5, label='Sigmoid (smooth)')

# Step function
step = np.where(x_range >= 0, 1, 0)
ax.plot(x_range, step, 'r-', linewidth=2.5, label='Step function (hard threshold)', alpha=0.7)

ax.set_title('Sigmoid vs Step Function', fontsize=12, fontweight='bold')
ax.set_xlabel('x', fontsize=11)
ax.set_ylabel('Output', fontsize=11)
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3)
ax.set_ylim(-0.1, 1.1)

plt.tight_layout()
plt.savefig('sigmoid_comprehensive.png', dpi=150, bbox_inches='tight')
plt.show()

# Additional mathematical properties demonstration
print("\n" + "="*60)
print("MATHEMATICAL PROPERTIES VERIFICATION")
print("="*60)

# Property 1: σ(-x) = 1 - σ(x)
x_test = 3
print(f"\n1. Symmetry: σ(-x) = 1 - σ(x)")
print(f"   σ({-x_test}) = {sigmoid(-x_test):.8f}")
print(f"   1 - σ({x_test}) = {1 - sigmoid(x_test):.8f}")
print(f"   Equal? {np.isclose(sigmoid(-x_test), 1 - sigmoid(x_test))}")

# Property 2: Derivative
print(f"\n2. Derivative: σ'(x) = σ(x)(1 - σ(x))")
x_test = 0
sig_val = sigmoid(x_test)
derivative_val = sig_val * (1 - sig_val)
print(f"   At x={x_test}: σ'({x_test}) = {derivative_val:.8f}")
print(f"   Maximum derivative is at x=0: {sigmoid(0) * (1 - sigmoid(0)):.8f}")

# Property 3: Limits
print(f"\n3. Limits:")
print(f"   lim(x→∞) σ(x) = {sigmoid(100):.10f} ≈ 1")
print(f"   lim(x→-∞) σ(x) = {sigmoid(-100):.10f} ≈ 0")

# Property 4: Inverse (logit)
print(f"\n4. Inverse function (logit): logit(p) = ln(p/(1-p))")
p_test = 0.8
logit = np.log(p_test / (1 - p_test))
recovered = sigmoid(logit)
```

### Exercise 2: Logistic Regression Probability Calculation

Use the sigmoid function to calculate class probabilities.

```python
feature1 = 1.5
feature2 = -0.8
bias = 0.5
coef1 = 0.8
coef2 = -0.3

# Calculate the linear combination z
# z = w₁·x₁ + w₂·x₂ + b
z = coef1 * feature1 + coef2 * feature2 + bias

# Use sigmoid to calculate probability of class 1
# P(y=1|x) = σ(z)
probability = sigmoid(z)

print(f"\nStep-by-step calculation:")
print(f"Feature 1 (x₁): {feature1}")
print(f"Feature 2 (x₂): {feature2}")
print(f"Coefficient 1 (w₁): {coef1}")
print(f"Coefficient 2 (w₂): {coef2}")
print(f"Bias (b): {bias}")
print(f"\nLinear combination z = w₁·x₁ + w₂·x₂ + b")
print(f"z = {coef1}·{feature1} + {coef2}·{feature2} + {bias}")
print(f"z = {coef1 * feature1} + {coef2 * feature2} + {bias}")
print(f"z = {z:.4f}")
print(f"\nProbability P(y=1|x) = σ(z) = σ({z:.4f})")
print(f"Probability of class 1: {probability:.4f}")
print(f"Probability of class 0: {1 - probability:.4f}")

# Based on a threshold of 0.5, make a prediction
threshold = 0.5
prediction = 1 if probability >= threshold else 0

print(f"\nDecision (threshold = {threshold}):")
print(f"Predicted class: {prediction}")
if prediction == 1:
    print(f"→ Class 1 (Positive) because P(y=1) = {probability:.4f} >= {threshold}")
else:
    print(f"→ Class 0 (Negative) because P(y=1) = {probability:.4f} < {threshold}")

# Create a function that takes features, coefficients, and bias
# and returns both probability and prediction
def predict_probability(features, coefficients, bias, threshold=0.5):
    """
    Calculate probability and prediction using sigmoid function

    Parameters:
    -----------
    features : array-like
        Input features [x₁, x₂, ..., xₙ]
    coefficients : array-like
        Model weights [w₁, w₂, ..., wₙ]
    bias : float
        Model bias term (intercept)
    threshold : float, default=0.5
        Decision threshold for classification

    Returns:
    --------
    probability : float
        Probability of class 1, P(y=1|x)
    prediction : int
        Predicted class (0 or 1)
    """
    # Convert to numpy arrays for easier computation
    features = np.array(features)
    coefficients = np.array(coefficients)

    # Calculate linear combination: z = w·x + b
    z = np.dot(coefficients, features) + bias

    # Apply sigmoid to get probability
    probability = sigmoid(z)

    # Make prediction based on threshold
    prediction = 1 if probability >= threshold else 0

    return probability, prediction

print("\n" + "="*70)
print("TESTING THE PREDICTION FUNCTION")
print("="*70)

# Test the function
test_features = [1.5, -0.8]
test_coefficients = [0.8, -0.3]
test_bias = 0.5

prob, pred = predict_probability(test_features, test_coefficients, test_bias)
print(f"\nTest - Probability: {prob:.4f}, Prediction: {pred}")

# Test with multiple examples
print("\n" + "="*70)
print("MULTIPLE TEST CASES")
print("="*70)

test_cases = [
    ([1.5, -0.8], "Original example"),
    ([2.0, 1.0], "Both features positive"),
    ([-1.0, -1.0], "Both features negative"),
    ([0.0, 0.0], "Zero features"),
    ([5.0, 2.0], "Large positive features"),
    ([-5.0, -2.0], "Large negative features"),
]

print(f"{'Features':<20} {'z':<10} {'P(y=1)':<10} {'Prediction':<12} {'Description'}")
print("-" * 80)

for features, description in test_cases:
    prob, pred = predict_probability(features, test_coefficients, test_bias)
    z = np.dot(test_coefficients, features) + test_bias
    print(f"{str(features):<20} {z:<10.4f} {prob:<10.4f} {pred:<12} {description}")

# Visualize decision boundary in 2D
print("\n" + "="*70)
print("VISUALIZING DECISION BOUNDARY")
print("="*70)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Decision boundary and regions
ax = axes[0]

# Create a mesh grid
x1_range = np.linspace(-3, 3, 300)
x2_range = np.linspace(-3, 3, 300)
X1, X2 = np.meshgrid(x1_range, x2_range)

# Calculate probabilities for each point
Z = np.zeros_like(X1)
for i in range(X1.shape[0]):
    for j in range(X1.shape[1]):
        features = [X1[i, j], X2[i, j]]
        Z[i, j], _ = predict_probability(features, test_coefficients, test_bias)

# Plot contour
contour = ax.contourf(X1, X2, Z, levels=20, cmap='RdYlBu_r', alpha=0.8)
plt.colorbar(contour, ax=ax, label='P(y=1|x)')

# Plot decision boundary (where P(y=1) = 0.5)
ax.contour(X1, X2, Z, levels=[0.5], colors='black', linewidths=3, linestyles='--')

# Plot test points
for features, description in test_cases[:4]:  # Plot first 4 test cases
    prob, pred = predict_probability(features, test_coefficients, test_bias)
    color = 'blue' if pred == 1 else 'red'
    marker = 'o' if pred == 1 else 's'
    ax.scatter(features[0], features[1], c=color, s=200, marker=marker,
              edgecolors='black', linewidths=2, zorder=5)
    ax.annotate(f'P={prob:.2f}', (features[0], features[1]),
               xytext=(5, 5), textcoords='offset points', fontsize=9,
               bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

ax.set_xlabel('Feature 1 (x₁)', fontsize=12, fontweight='bold')
ax.set_ylabel('Feature 2 (x₂)', fontsize=12, fontweight='bold')
ax.set_title('Decision Boundary and Probability Regions', fontsize=13, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(['Decision Boundary (P=0.5)', 'Class 1', 'Class 0'], loc='upper right')

# Plot 2: Probability surface in 3D perspective
ax = axes[1]

# Sample points along a line
x1_line = np.linspace(-3, 3, 100)
x2_line = np.linspace(-3, 3, 100)

# Calculate probabilities along x1 axis (x2=0)
probs_x1 = []
for x1 in x1_line:
    prob, _ = predict_probability([x1, 0], test_coefficients, test_bias)
    probs_x1.append(prob)

# Calculate probabilities along x2 axis (x1=0)
probs_x2 = []
for x2 in x2_line:
    prob, _ = predict_probability([0, x2], test_coefficients, test_bias)
    probs_x2.append(prob)

ax.plot(x1_line, probs_x1, 'b-', linewidth=2.5, label='Along x₁ axis (x₂=0)')
ax.plot(x2_line, probs_x2, 'r-', linewidth=2.5, label='Along x₂ axis (x₁=0)')
ax.axhline(y=0.5, color='black', linestyle='--', linewidth=2, label='Decision threshold')
ax.fill_between(x1_line, 0, 0.5, alpha=0.2, color='red', label='Class 0 region')
ax.fill_between(x1_line, 0.5, 1, alpha=0.2, color='blue', label='Class 1 region')

ax.set_xlabel('Feature value', fontsize=12, fontweight='bold')
ax.set_ylabel('P(y=1|x)', fontsize=12, fontweight='bold')
ax.set_title('Probability Along Feature Axes', fontsize=13, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=10)
ax.set_ylim(-0.05, 1.05)

plt.tight_layout()
plt.savefig('logistic_regression_probability.png', dpi=150, bbox_inches='tight')
plt.show()

# Additional analysis: Effect of threshold
print("\n" + "="*70)
print("EFFECT OF DIFFERENT THRESHOLDS")
print("="*70)

test_feature = [1.5, -0.8]
prob, _ = predict_probability(test_feature, test_coefficients, test_bias)

thresholds = [0.3, 0.4, 0.5, 0.6, 0.7]
print(f"\nFor features {test_feature} with P(y=1) = {prob:.4f}:\n")
print(f"{'Threshold':<12} {'Prediction':<12} {'Reasoning'}")
print("-" * 60)

for thresh in thresholds:
    pred = 1 if prob >= thresh else 0
    reasoning = f"{prob:.4f} >= {thresh}" if pred == 1 else f"{prob:.4f} < {thresh}"
    print(f"{thresh:<12} {pred:<12} {reasoning}")

# Show relationship between z and probability
print("\n" + "="*70)
print("RELATIONSHIP BETWEEN z AND PROBABILITY")
print("="*70)

fig, ax = plt.subplots(figsize=(12, 6))

z_values = np.linspace(-6, 6, 100)
prob_values = sigmoid(z_values)

ax.plot(z_values, prob_values, 'b-', linewidth=3, label='P(y=1) = σ(z)')
ax.axhline(y=0.5, color='red', linestyle='--', linewidth=2, label='Decision boundary (P=0.5)')
ax.axvline(x=0, color='red', linestyle='--', linewidth=2, alpha=0.5)

# Mark our example
z_example = np.dot(test_coefficients, test_features) + test_bias
prob_example = sigmoid(z_example)
ax.plot(z_example, prob_example, 'go', markersize=15, label=f'Our example: z={z_example:.2f}', zorder=5)
ax.annotate(f'z={z_example:.2f}\nP(y=1)={prob_example:.4f}',
           xy=(z_example, prob_example), xytext=(z_example+1, prob_example-0.2),
           arrowprops=dict(arrowstyle='->', color='green', lw=2),
           fontsize=11, bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.8))

ax.fill_between(z_values, 0, 0.5, alpha=0.2, color='red', label='Predict Class 0')
ax.fill_between(z_values, 0.5, 1, alpha=0.2, color='blue', label='Predict Class 1')

ax.set_xlabel('z (linear combination)', fontsize=13, fontweight='bold')
ax.set_ylabel('P(y=1|x)', fontsize=13, fontweight='bold')
ax.set_title('Sigmoid Function: Converting Linear Score to Probability',
            fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11)
ax.set_ylim(-0.05, 1.05)

plt.tight_layout()
plt.savefig('z_to_probability.png', dpi=150, bbox_inches='tight')
plt.show()
```

### Exercise 3: Confusion Matrix Implementation

Implement a confusion matrix calculation from scratch.

```python
# Sample true labels and predictions
y_true = [0, 1, 0, 1, 1, 0, 1, 0, 0, 1]
y_pred = [0, 1, 1, 1, 0, 0, 1, 0, 1, 1]

def calculate_confusion_matrix(y_true, y_pred):
    """
    Calculate confusion matrix components from scratch

    Parameters:
    -----------
    y_true : list or array
        True labels (ground truth)
    y_pred : list or array
        Predicted labels

    Returns:
    --------
    TP, TN, FP, FN : int
        True Positives, True Negatives, False Positives, False Negatives
    """
    TP = TN = FP = FN = 0

    # Iterate through each prediction
    for true_label, pred_label in zip(y_true, y_pred):
        if true_label == 1 and pred_label == 1:
            TP += 1  # Correctly predicted positive
        elif true_label == 0 and pred_label == 0:
            TN += 1  # Correctly predicted negative
        elif true_label == 0 and pred_label == 1:
            FP += 1  # Incorrectly predicted as positive (False Alarm)
        elif true_label == 1 and pred_label == 0:
            FN += 1  # Incorrectly predicted as negative (Miss)

    return TP, TN, FP, FN

# Calculate confusion matrix
TP, TN, FP, FN = calculate_confusion_matrix(y_true, y_pred)

print("\n" + "="*70)
print("CONFUSION MATRIX COMPONENTS")
print("="*70)
print(f"True Positives (TP):  {TP} - Correctly predicted as positive")
print(f"True Negatives (TN):  {TN} - Correctly predicted as negative")
print(f"False Positives (FP): {FP} - Incorrectly predicted as positive (Type I Error)")
print(f"False Negatives (FN): {FN} - Incorrectly predicted as negative (Type II Error)")
print(f"\nTotal predictions: {TP + TN + FP + FN}")
print(f"Correct predictions: {TP + TN}")
print(f"Incorrect predictions: {FP + FN}")

# Show detailed breakdown
print("\n" + "="*70)
print("DETAILED BREAKDOWN")
print("="*70)
print(f"{'Index':<8} {'True':<8} {'Pred':<8} {'Result':<20} {'Category'}")
print("-" * 70)

for i, (true_label, pred_label) in enumerate(zip(y_true, y_pred)):
    if true_label == 1 and pred_label == 1:
        result = "✓ Correct"
        category = "True Positive (TP)"
    elif true_label == 0 and pred_label == 0:
        result = "✓ Correct"
        category = "True Negative (TN)"
    elif true_label == 0 and pred_label == 1:
        result = "✗ Incorrect"
        category = "False Positive (FP)"
    else:  # true_label == 1 and pred_label == 0
        result = "✗ Incorrect"
        category = "False Negative (FN)"

    print(f"{i:<8} {true_label:<8} {pred_label:<8} {result:<20} {category}")

# Calculate metrics from confusion matrix
def calculate_metrics(TP, TN, FP, FN):
    """Calculate various classification metrics"""

    # Accuracy: Overall correctness
    accuracy = (TP + TN) / (TP + TN + FP + FN) if (TP + TN + FP + FN) > 0 else 0

    # Precision: Of all predicted positives, how many were correct?
    precision = TP / (TP + FP) if (TP + FP) > 0 else 0

    # Recall (Sensitivity/True Positive Rate): Of all actual positives, how many did we find?
    recall = TP / (TP + FN) if (TP + FN) > 0 else 0

    # Specificity (True Negative Rate): Of all actual negatives, how many did we correctly identify?
    specificity = TN / (TN + FP) if (TN + FP) > 0 else 0

    # F1 Score: Harmonic mean of precision and recall
    f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

    # False Positive Rate
    fpr = FP / (FP + TN) if (FP + TN) > 0 else 0

    # False Negative Rate
    fnr = FN / (FN + TP) if (FN + TP) > 0 else 0

    return {
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'specificity': specificity,
        'f1_score': f1_score,
        'fpr': fpr,
        'fnr': fnr
    }

metrics = calculate_metrics(TP, TN, FP, FN)

print("\n" + "="*70)
print("CLASSIFICATION METRICS")
print("="*70)
print(f"Accuracy:    {metrics['accuracy']:.4f} ({metrics['accuracy']*100:.2f}%)")
print(f"             → Overall correctness: (TP+TN)/(TP+TN+FP+FN)")
print(f"Precision:   {metrics['precision']:.4f} ({metrics['precision']*100:.2f}%)")
print(f"             → Of predicted positives, % correct: TP/(TP+FP)")
print(f"Recall:      {metrics['recall']:.4f} ({metrics['recall']*100:.2f}%)")
print(f"             → Of actual positives, % found: TP/(TP+FN)")
print(f"Specificity: {metrics['specificity']:.4f} ({metrics['specificity']*100:.2f}%)")
print(f"             → Of actual negatives, % found: TN/(TN+FP)")
print(f"F1-Score:    {metrics['f1_score']:.4f}")
print(f"             → Harmonic mean of precision & recall")
print(f"FPR:         {metrics['fpr']:.4f} ({metrics['fpr']*100:.2f}%)")
print(f"             → False Positive Rate: FP/(FP+TN)")
print(f"FNR:         {metrics['fnr']:.4f} ({metrics['fnr']*100:.2f}%)")
print(f"             → False Negative Rate: FN/(FN+TP)")

# ============================================================================
# VISUALIZATIONS
# ============================================================================

fig = plt.figure(figsize=(18, 12))

# Plot 1: Basic Confusion Matrix
ax1 = plt.subplot(2, 3, 1)
conf_matrix = np.array([[TN, FP],
                        [FN, TP]])

sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Predicted 0', 'Predicted 1'],
            yticklabels=['Actual 0', 'Actual 1'],
            cbar_kws={'label': 'Count'},
            linewidths=2, linecolor='black',
            annot_kws={'size': 16, 'weight': 'bold'})
ax1.set_title('Confusion Matrix', fontsize=14, fontweight='bold', pad=15)
ax1.set_ylabel('True Label', fontsize=12, fontweight='bold')
ax1.set_xlabel('Predicted Label', fontsize=12, fontweight='bold')

# Plot 2: Annotated Confusion Matrix with Labels
ax2 = plt.subplot(2, 3, 2)
labels = np.array([[f'TN\n{TN}', f'FP\n{FP}'],
                   [f'FN\n{FN}', f'TP\n{TP}']])
colors = np.array([[TN, FP], [FN, TP]])

sns.heatmap(colors, annot=labels, fmt='', cmap='RdYlGn',
            xticklabels=['Predicted Negative (0)', 'Predicted Positive (1)'],
            yticklabels=['Actual Negative (0)', 'Actual Positive (1)'],
            cbar=False,
            linewidths=3, linecolor='black',
            annot_kws={'size': 14, 'weight': 'bold'})
ax2.set_title('Confusion Matrix with Labels', fontsize=14, fontweight='bold', pad=15)
ax2.set_ylabel('True Label', fontsize=12, fontweight='bold')
ax2.set_xlabel('Predicted Label', fontsize=12, fontweight='bold')

# Plot 3: Normalized Confusion Matrix (Percentages)
ax3 = plt.subplot(2, 3, 3)
total = TP + TN + FP + FN
conf_matrix_norm = conf_matrix / total * 100

sns.heatmap(conf_matrix_norm, annot=True, fmt='.1f', cmap='YlOrRd',
            xticklabels=['Predicted 0', 'Predicted 1'],
            yticklabels=['Actual 0', 'Actual 1'],
            cbar_kws={'label': 'Percentage (%)'},
            linewidths=2, linecolor='black',
            annot_kws={'size': 14, 'weight': 'bold'})
ax3.set_title('Normalized Confusion Matrix (%)', fontsize=14, fontweight='bold', pad=15)
ax3.set_ylabel('True Label', fontsize=12, fontweight='bold')
ax3.set_xlabel('Predicted Label', fontsize=12, fontweight='bold')

# Plot 4: Metrics Bar Chart
ax4 = plt.subplot(2, 3, 4)
metric_names = ['Accuracy', 'Precision', 'Recall', 'Specificity', 'F1-Score']
metric_values = [metrics['accuracy'], metrics['precision'], metrics['recall'],
                metrics['specificity'], metrics['f1_score']]
colors_bar = ['#2ecc71', '#3498db', '#e74c3c', '#f39c12', '#9b59b6']

bars = ax4.barh(metric_names, metric_values, color=colors_bar, edgecolor='black', linewidth=2)
ax4.set_xlim(0, 1)
ax4.set_xlabel('Score', fontsize=12, fontweight='bold')
ax4.set_title('Classification Metrics', fontsize=14, fontweight='bold')
ax4.grid(axis='x', alpha=0.3)

# Add value labels on bars
for i, (bar, value) in enumerate(zip(bars, metric_values)):
    ax4.text(value + 0.02, i, f'{value:.3f}', va='center', fontsize=11, fontweight='bold')

# Plot 5: Prediction Visualization
ax5 = plt.subplot(2, 3, 5)
x_pos = np.arange(len(y_true))
colors_pred = ['green' if t == p else 'red' for t, p in zip(y_true, y_pred)]
markers = ['o' if t == 1 else 's' for t in y_true]

for i, (true_label, pred_label, color, marker) in enumerate(zip(y_true, y_pred, colors_pred, markers)):
    ax5.scatter(i, true_label, c='blue', s=200, marker=marker,
               edgecolors='black', linewidths=2, alpha=0.6, label='True' if i == 0 else '')
    ax5.scatter(i, pred_label, c=color, s=150, marker='x',
               linewidths=3, label='Pred' if i == 0 else '')
    ax5.plot([i, i], [true_label, pred_label], 'k--', alpha=0.3, linewidth=1)

ax5.set_xlabel('Sample Index', fontsize=12, fontweight='bold')
ax5.set_ylabel('Label', fontsize=12, fontweight='bold')
ax5.set_title('True vs Predicted Labels', fontsize=14, fontweight='bold')
ax5.set_yticks([0, 1])
ax5.set_yticklabels(['Negative (0)', 'Positive (1)'])
ax5.grid(axis='y', alpha=0.3)
ax5.legend(['True (circle=1, square=0)', 'Pred (green=correct, red=wrong)'],
          fontsize=9, loc='upper right')

# Plot 6: Error Types Distribution
ax6 = plt.subplot(2, 3, 6)
categories = ['True\nPositive\n(TP)', 'True\nNegative\n(TN)',
             'False\nPositive\n(FP)', 'False\nNegative\n(FN)']
values = [TP, TN, FP, FN]
colors_dist = ['#2ecc71', '#3498db', '#e74c3c', '#f39c12']

bars = ax6.bar(categories, values, color=colors_dist, edgecolor='black', linewidth=2)
ax6.set_ylabel('Count', fontsize=12, fontweight='bold')
ax6.set_title('Distribution of Prediction Types', fontsize=14, fontweight='bold')
ax6.grid(axis='y', alpha=0.3)

# Add value labels on bars
for bar, value in zip(bars, values):
    height = bar.get_height()
    ax6.text(bar.get_x() + bar.get_width()/2., height,
            f'{int(value)}', ha='center', va='bottom', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('confusion_matrix_comprehensive.png', dpi=150, bbox_inches='tight')
plt.show()

# Additional: Create a detailed explanation diagram
fig, ax = plt.subplots(figsize=(14, 10))
ax.axis('off')

# Title
ax.text(0.5, 0.95, 'Understanding the Confusion Matrix',
        ha='center', fontsize=18, fontweight='bold', transform=ax.transAxes)

# Draw confusion matrix structure
from matplotlib.patches import Rectangle

# Define positions
left, bottom = 0.2, 0.5
width, height = 0.25, 0.25

# TN box
rect_tn = Rectangle((left, bottom + height), width, height,
                    linewidth=3, edgecolor='blue', facecolor='lightblue', alpha=0.5)
ax.add_patch(rect_tn)
ax.text(left + width/2, bottom + height + height/2, f'TN = {TN}\nTrue Negative\nCorrect ✓',
        ha='center', va='center', fontsize=12, fontweight='bold')

# FP box
rect_fp = Rectangle((left + width, bottom + height), width, height,
                    linewidth=3, edgecolor='red', facecolor='lightcoral', alpha=0.5)
ax.add_patch(rect_fp)
ax.text(left + width + width/2, bottom + height + height/2, f'FP = {FP}\nFalse Positive\nType I Error ✗',
        ha='center', va='center', fontsize=12, fontweight='bold')

# FN box
rect_fn = Rectangle((left, bottom), width, height,
                    linewidth=3, edgecolor='orange', facecolor='lightyellow', alpha=0.5)
ax.add_patch(rect_fn)
ax.text(left + width/2, bottom + height/2, f'FN = {FN}\nFalse Negative\nType II Error ✗',
        ha='center', va='center', fontsize=12, fontweight='bold')

# TP box
rect_tp = Rectangle((left + width, bottom), width, height,
                    linewidth=3, edgecolor='green', facecolor='lightgreen', alpha=0.5)
ax.add_patch(rect_tp)
ax.text(left + width + width/2, bottom + height/2, f'TP = {TP}\nTrue Positive\nCorrect ✓',
        ha='center', va='center', fontsize=12, fontweight='bold')

# Labels
ax.text(left + width, bottom + 2*height + 0.05, 'Predicted',
        ha='center', fontsize=13, fontweight='bold')
ax.text(left + width/2, bottom + 2*height + 0.02, 'Negative (0)',
        ha='center', fontsize=11)
ax.text(left + 1.5*width, bottom + 2*height + 0.02, 'Positive (1)',
        ha='center', fontsize=11)

ax.text(left - 0.08, bottom + height, 'Actual',
        ha='center', va='center', fontsize=13, fontweight='bold', rotation=90)
ax.text(left - 0.03, bottom + 1.5*height, 'Negative (0)',
        ha='center', va='center', fontsize=11, rotation=90)
ax.text(left - 0.03, bottom + 0.5*height, 'Positive (1)',
        ha='center', va='center', fontsize=11, rotation=90)

# Formulas
formula_y = 0.35
ax.text(0.1, formula_y, 'Key Formulas:', fontsize=14, fontweight='bold',
        transform=ax.transAxes)
ax.text(0.1, formula_y - 0.05, f'• Accuracy = (TP + TN) / Total = {metrics["accuracy"]:.3f}',
        fontsize=11, transform=ax.transAxes)
ax.text(0.1, formula_y - 0.10, f'• Precision = TP / (TP + FP) = {metrics["precision"]:.3f}',
        fontsize=11, transform=ax.transAxes)
ax.text(0.1, formula_y - 0.15, f'• Recall = TP / (TP + FN) = {metrics["recall"]:.3f}',
        fontsize=11, transform=ax.transAxes)
ax.text(0.1, formula_y - 0.20, f'• F1-Score = 2 × (Precision × Recall) / (Precision + Recall) = {metrics["f1_score"]:.3f}',
        fontsize=11, transform=ax.transAxes)

# Interpretations
interp_y = 0.10
ax.text(0.1, interp_y, 'Interpretations:', fontsize=14, fontweight='bold',
        transform=ax.transAxes)
ax.text(0.1, interp_y - 0.04, '• TP: Model correctly identified positive cases',
        fontsize=10, transform=ax.transAxes)
ax.text(0.1, interp_y - 0.07, '• TN: Model correctly identified negative cases',
        fontsize=10, transform=ax.transAxes)
ax.text(0.1, interp_y - 0.10, '• FP: Model incorrectly predicted positive (False Alarm)',
        fontsize=10, transform=ax.transAxes)
ax.text(0.1, interp_y - 0.13, '• FN: Model missed actual positive cases (Miss)',
        fontsize=10, transform=ax.transAxes)

plt.tight_layout()
plt.savefig('confusion_matrix_explanation.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
def calculate_confusion_matrix(y_true, y_pred):
    """Calculate TP, TN, FP, FN"""
    TP = TN = FP = FN = 0

    for true_label, pred_label in zip(y_true, y_pred):
        if true_label == 1 and pred_label == 1:
            TP += 1
        elif true_label == 0 and pred_label == 0:
            TN += 1
        elif true_label == 0 and pred_label == 1:
            FP += 1
        elif true_label == 1 and pred_label == 0:
            FN += 1

    return TP, TN, FP, FN

# Get confusion matrix components
TP, TN, FP, FN = calculate_confusion_matrix(y_true, y_pred)

def calculate_metrics(TP, TN, FP, FN):
    """
    Calculate classification metrics from confusion matrix components

    Parameters:
    -----------
    TP : int - True Positives
    TN : int - True Negatives
    FP : int - False Positives
    FN : int - False Negatives

    Returns:
    --------
    accuracy, precision, recall, f1 : float
        The four main classification metrics
    """

    # Accuracy: What fraction of predictions were correct?
    # Formula: (TP + TN) / (TP + TN + FP + FN)
    # Interpretation: Overall correctness of the model
    total = TP + TN + FP + FN
    accuracy = (TP + TN) / total if total > 0 else 0

    # Precision: Of all positive predictions, how many were actually positive?
    # Formula: TP / (TP + FP)
    # Interpretation: Quality of positive predictions (minimizes false alarms)
    precision = TP / (TP + FP) if (TP + FP) > 0 else 0

    # Recall (Sensitivity/True Positive Rate): Of all actual positives, how many did we find?
    # Formula: TP / (TP + FN)
    # Interpretation: Coverage of positive class (minimizes misses)
    recall = TP / (TP + FN) if (TP + FN) > 0 else 0

    # F1-Score: Harmonic mean of precision and recall
    # Formula: 2 * (precision * recall) / (precision + recall)
    # Interpretation: Balance between precision and recall
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

    return accuracy, precision, recall, f1

# Calculate metrics
accuracy, precision, recall, f1 = calculate_metrics(TP, TN, FP, FN)

print("\nClassification Metrics:")
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")
```
